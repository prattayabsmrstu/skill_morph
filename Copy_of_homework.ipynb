```python
def celsius_to_fahrenheit(celsius):
    fahrenheit = (celsius * 9/5) + 32
    return fahrenheit

# Example usage:
temp_c = 25
temp_f = celsius_to_fahrenheit(temp_c)
print(f"{temp_c}°C is equal to {temp_f}°F")

```

```python
def rectangle_area(length=10, width=5):
    area = length * width
    return area

# Example usage:
print("Default area:", rectangle_area())             # Uses default values: 10 × 5
print("Custom area:", rectangle_area(7, 3))          # Uses custom values: 7 × 3
print("Length only:", rectangle_area(length=8))      # Uses 8 × default width (5)
print("Width only:", rectangle_area(width=4))        # Uses default length (10) × 4
```
