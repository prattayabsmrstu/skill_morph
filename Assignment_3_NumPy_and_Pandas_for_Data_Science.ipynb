```python
#Task 3.2: Basic Groupby Analysis
# Comprehensive groupby analysis
# Your tasks:

# 1. Group by 'gender' and calculate:
#    - Average scores in all subjects
#    - Count of students

# 2. Group by 'test preparation course' and show:
#    - Mean scores for each subject
#    - Show which subject benefits most from test prep

# 3. Group by 'parental level of education' and calculate:
#    - Average math score for each education level
#    - Rank education levels by math performance

# 4. Group by 'lunch' type and show:
#    - Average scores for each subject
#    - Count of students in each lunch category

# Your code here:
#task 1
avg_scores = df.groupby('gender')[['math score', 'reading score', 'writing score']].mean()
print(avg_scores)

student_counts = df['gender'].value_counts()
print(student_counts)

#task 2
prep_scores = df.groupby('test preparation course')[['math score', 'reading score', 'writing score']].mean()
score_diff = prep_scores.loc['completed'] - prep_scores.loc['none']
most_improved_subject = score_diff.idxmax()

print(" Mean Scores by Test Preparation Course:")
print(prep_scores)

print("\n Score Improvements from Test Prep:")
print(score_diff)

print(f"\n Subject that benefits most from test prep: {most_improved_subject}")

#3
math_by_education = df.groupby('parental level of education')['math score'].mean()
ranked_math = math_by_education.sort_values(ascending=False)

# Display results
print("Average Math Score by Parental Education Level:")
print(math_by_education)

print("\n Ranked Education Levels by Math Performance:")
print(ranked_math)

#4
# Group by lunch type and calculate average scores
avg_scores_by_lunch = df.groupby('lunch')[['math score', 'reading score', 'writing score']].mean()

# Count of students in each lunch category
student_count_by_lunch = df['lunch'].value_counts()

# Display results
print("Average Scores by Lunch Type:")
print(avg_scores_by_lunch)

print("\nStudent Count by Lunch Type:")
print(student_count_by_lunch)
```
